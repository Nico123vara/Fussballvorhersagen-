# Feature Engineering — Fussball Vorhersagen
Wir laden die Spieldaten aus der Datenbank und berechnen Features für das ML-Modell.

```python
import pandas as pd
import psycopg2
import matplotlib.pyplot as plt
import seaborn as sns

DB_CONFIG = {
    'host': 'localhost',
    'port': 5432,
    'dbname': 'fussball',
    'user': 'fussball',
    'password': 'fussball_pw'
}

conn = psycopg2.connect(**DB_CONFIG)
print('Verbindung erfolgreich!')
```

## 1. Daten aus DB laden

```python
query = """
    SELECT 
        m.id,
        m.utc_date,
        m.matchday,
        m.status,
        m.season_id,
        m.home_team_id,
        m.away_team_id,
        t1.name AS home_team,
        t2.name AS away_team,
        m.home_score_fulltime,
        m.away_score_fulltime,
        m.winner
    FROM matches m
    JOIN teams t1 ON m.home_team_id = t1.id
    JOIN teams t2 ON m.away_team_id = t2.id
    WHERE m.status = 'FINISHED'
    ORDER BY m.utc_date ASC
"""

df = pd.read_sql(query, conn)
print(f'Spiele geladen: {len(df)}')
df.head(10)
```

## 2. Daten kennenlernen

```python
print('Ergebnis-Verteilung:')
print(df['winner'].value_counts())
print()
print(f'Heimsiege:    {(df["winner"] == "HOME_TEAM").sum()} ({(df["winner"] == "HOME_TEAM").mean()*100:.1f}%)')
print(f'Unentschieden: {(df["winner"] == "DRAW").sum()} ({(df["winner"] == "DRAW").mean()*100:.1f}%)')
print(f'Auswärtssiege: {(df["winner"] == "AWAY_TEAM").sum()} ({(df["winner"] == "AWAY_TEAM").mean()*100:.1f}%)')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Ergebnis-Verteilung
df['winner'].value_counts().plot(kind='bar', ax=axes[0], color=['#2a78d6', '#1baf7a', '#e34948'])
axes[0].set_title('Ergebnis-Verteilung')
axes[0].set_xlabel('')
axes[0].tick_params(rotation=0)

# Tore pro Spiel
df['total_tore'] = df['home_score_fulltime'] + df['away_score_fulltime']
df['total_tore'].hist(bins=10, ax=axes[1], color='#2a78d6')
axes[1].set_title('Tore pro Spiel')
axes[1].set_xlabel('Anzahl Tore')

plt.tight_layout()
plt.show()
```

## 3. Feature Engineering
Wir berechnen für jedes Spiel Features die **vor dem Spiel bekannt waren**.

```python
def get_form(df, team_id, date, n=5):
    """
    Berechnet die Punkte der letzten n Spiele eines Teams vor einem bestimmten Datum.
    Sieg = 3 Punkte, Unentschieden = 1, Niederlage = 0
    """
    vergangene = df[
        ((df['home_team_id'] == team_id) | (df['away_team_id'] == team_id)) &
        (df['utc_date'] < date)
    ].tail(n)
    
    punkte = 0
    for _, spiel in vergangene.iterrows():
        if spiel['home_team_id'] == team_id:
            if spiel['winner'] == 'HOME_TEAM': punkte += 3
            elif spiel['winner'] == 'DRAW': punkte += 1
        else:
            if spiel['winner'] == 'AWAY_TEAM': punkte += 3
            elif spiel['winner'] == 'DRAW': punkte += 1
    
    return punkte


def get_head_to_head(df, home_id, away_id, date, n=5):
    """
    Head-to-Head: Wie oft hat das Heimteam gegen das Auswärtsteam gewonnen
    in den letzten n Direktbegegnungen.
    """
    h2h = df[
        ((df['home_team_id'] == home_id) & (df['away_team_id'] == away_id) |
         (df['home_team_id'] == away_id) & (df['away_team_id'] == home_id)) &
        (df['utc_date'] < date)
    ].tail(n)
    
    if len(h2h) == 0:
        return 0
    
    siege = 0
    for _, spiel in h2h.iterrows():
        if spiel['home_team_id'] == home_id and spiel['winner'] == 'HOME_TEAM':
            siege += 1
        elif spiel['away_team_id'] == home_id and spiel['winner'] == 'AWAY_TEAM':
            siege += 1
    
    return siege / len(h2h)


def get_heimquote(df, team_id, date):
    """
    Heimquote: Anteil Heimsiege des Teams in der aktuellen Saison.
    """
    heimspiele = df[
        (df['home_team_id'] == team_id) &
        (df['utc_date'] < date)
    ]
    
    if len(heimspiele) == 0:
        return 0.5  # Standardwert wenn keine Daten
    
    return (heimspiele['winner'] == 'HOME_TEAM').mean()


print('Funktionen definiert!')
```

```python
# Features für alle Spiele berechnen
# Achtung: dauert etwas da wir für jedes Spiel die Historie durchsuchen

print('Berechne Features...')

features = []
for _, row in df.iterrows():
    home_form  = get_form(df, row['home_team_id'], row['utc_date'])
    away_form  = get_form(df, row['away_team_id'], row['utc_date'])
    h2h        = get_head_to_head(df, row['home_team_id'], row['away_team_id'], row['utc_date'])
    heimquote  = get_heimquote(df, row['home_team_id'], row['utc_date'])
    
    features.append({
        'match_id':       row['id'],
        'home_team':      row['home_team'],
        'away_team':      row['away_team'],
        'home_form':      home_form,
        'away_form':      away_form,
        'form_diff':      home_form - away_form,
        'h2h':            h2h,
        'heimquote':      heimquote,
        'winner':         row['winner']
    })

df_features = pd.DataFrame(features)
print(f'Fertig! {len(df_features)} Spiele mit Features')
df_features.head(10)
```

## 4. Features anschauen

```python
# Statistiken der Features
df_features[['home_form', 'away_form', 'form_diff', 'h2h', 'heimquote']].describe()
```

```python
# Zusammenhang Form und Ergebnis
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df_features.groupby('winner')['home_form'].mean().plot(kind='bar', ax=axes[0], color='#2a78d6')
axes[0].set_title('Durchschnittliche Heimteam-Form nach Ergebnis')
axes[0].tick_params(rotation=0)

df_features.groupby('winner')['form_diff'].mean().plot(kind='bar', ax=axes[1], color='#1baf7a')
axes[1].set_title('Durchschnittliche Form-Differenz nach Ergebnis')
axes[1].tick_params(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Features speichern für ML-Modell
df_features.to_csv('data/features.csv', index=False)
print('Features gespeichert unter data/features.csv')
```
